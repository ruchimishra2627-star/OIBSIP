### Objective
The objective of this project is to clean and prepare a messy customer sales dataset using Python, Pandas, and NumPy.

The goal is to identify and handle missing values, duplicates, inconsistent data, incorrect data types, and outliers.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path = "Dataset/messy_customer_sales_data.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
```

```python
print("First 5 Rows:")
display(df.head())

print("Dataset Information:")
df.info()

print("Statistical Summary:")
display(df.describe(include="all"))
```

```python
quality_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100,
    "Data Type": df.dtypes,
    "Unique Values": df.nunique()
})

print("Initial Data Quality Report")
display(quality_report)

print("Total Duplicate Rows:", df.duplicated().sum())
```

```python
missing_data = df.isnull().sum()

missing_data = missing_data[missing_data > 0].sort_values(ascending=False)

print("Columns with Missing Values:")
display(missing_data)
```

```python
print("Actual Column Names:")
for col in df.columns:
    print(repr(col))
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace("'", "", regex=False)
    .str.replace(" ", "_")
)

print("Updated Column Names:")
print(df.columns.tolist())
```

```python
missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df)) * 100
})

display(missing_summary[missing_summary["Missing_Count"] > 0])
```

```python
print("Gender Values:")
print(df["Gender"].value_counts(dropna=False))

print("\nCity Values:")
print(df["City"].value_counts(dropna=False).head(15))

print("\nCountry Values:")
print(df["Country"].value_counts(dropna=False))

print("\nFeedback Score Values:")
print(df["Feedback_Score"].value_counts(dropna=False).sort_index())
```

```python
# Standardize Gender
df["Gender"] = df["Gender"].str.strip().str.lower()
df["Gender"] = df["Gender"].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female"
})

# Standardize City
df["City"] = df["City"].str.strip().str.title()

# Standardize Country
df["Country"] = df["Country"].str.strip().str.lower()
df["Country"] = df["Country"].replace({
    "india": "India",
    "ind": "India"
})

# Handle missing categorical values
for col in ["Gender", "City", "Country"]:
    df[col] = df[col].fillna("Unknown")

print("Gender:", df["Gender"].value_counts())
print("\nCity:", df["City"].value_counts())
print("\nCountry:", df["Country"].value_counts())
```

```python
df["Age"] = pd.to_numeric(
    df["Age"].astype("string").str.replace(" years", "", regex=False).str.strip(),
    errors="coerce"
)

print("Age Data Type:", df["Age"].dtype)
print("Missing Age Values:", df["Age"].isnull().sum())
print("Age Summary:")
display(df["Age"].describe())
```

```python
# Replace invalid ages with NaN
df.loc[(df["Age"] < 0) | (df["Age"] > 100), "Age"] = np.nan

print("Invalid ages removed.")
print("Missing Age Values:", df["Age"].isnull().sum())

# Fill missing ages with median
age_median = df["Age"].median()
df["Age"] = df["Age"].fillna(age_median)

print("Median Age Used:", age_median)
print("Remaining Missing Ages:", df["Age"].isnull().sum())
```

```python
print("Purchase Amount Data Type:", df["Purchase_Amount"].dtype)
print("Missing Values:", df["Purchase_Amount"].isnull().sum())

print("\nPurchase Amount Summary:")
display(df["Purchase_Amount"].describe())
```

```python
Q1 = df["Purchase_Amount"].quantile(0.25)
Q3 = df["Purchase_Amount"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Purchase_Amount"] < lower_limit) |
    (df["Purchase_Amount"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Total Outliers:", len(outliers))
```

```python
print("Outlier Values:")
display(outliers[["Purchase_Amount"]])
```

```python
# Replace negative purchase amounts with NaN
df.loc[df["Purchase_Amount"] < 0, "Purchase_Amount"] = np.nan

# Cap extreme outliers using IQR upper limit
df.loc[df["Purchase_Amount"] > upper_limit, "Purchase_Amount"] = upper_limit

# Fill missing values with median
purchase_median = df["Purchase_Amount"].median()
df["Purchase_Amount"] = df["Purchase_Amount"].fillna(purchase_median)

print("Median Used:", purchase_median)
print("Remaining Missing Values:", df["Purchase_Amount"].isnull().sum())
print("Minimum Amount:", df["Purchase_Amount"].min())
print("Maximum Amount:", df["Purchase_Amount"].max())
```

```python
print("Feedback Score Data Type:", df["Feedback_Score"].dtype)
print("Missing Values:", df["Feedback_Score"].isnull().sum())

print("Unique Values:")
print(sorted(df["Feedback_Score"].dropna().unique()))
```

```python
feedback_median = df["Feedback_Score"].median()

df["Feedback_Score"] = df["Feedback_Score"].fillna(feedback_median)

print("Median Used:", feedback_median)
print("Remaining Missing Values:", df["Feedback_Score"].isnull().sum())
```

```python
df["Signup_Date"] = pd.to_datetime(df["Signup_Date"], errors="coerce")
df["Last_Purchase_Date"] = pd.to_datetime(df["Last_Purchase_Date"], errors="coerce")

print("Signup Date Type:", df["Signup_Date"].dtype)
print("Last Purchase Date Type:", df["Last_Purchase_Date"].dtype)

print("\nMissing Signup Dates:", df["Signup_Date"].isnull().sum())
print("Missing Last Purchase Dates:", df["Last_Purchase_Date"].isnull().sum())
```

```python
print("Customer ID Data Type:", df["Customer_ID"].dtype)
print("Phone Number Data Type:", df["Phone_Number"].dtype)

print("\nMissing Customer IDs:", df["Customer_ID"].isnull().sum())
print("Missing Phone Numbers:", df["Phone_Number"].isnull().sum())

print("\nAll Column Data Types:")
display(df.dtypes)
```

```python
df["Phone_Number"] = df["Phone_Number"].astype("string")

print("Phone Number Data Type:", df["Phone_Number"].dtype)
print("\nPhone Number Lengths:")
print(df["Phone_Number"].str.len().value_counts())
```

```python
invalid_phones = df[df["Phone_Number"].str.len() != 10]

print("Total Invalid Phone Numbers:", len(invalid_phones))

display(invalid_phones[["Phone_Number"]].head(15))
```

```python
invalid_mask = df["Phone_Number"].str.len() != 10

df.loc[invalid_mask, "Phone_Number"] = pd.NA

print("Invalid Phone Numbers Handled:", invalid_mask.sum())
print("Remaining Invalid Numbers:",
      (df["Phone_Number"].str.len() != 10).sum())
print("Missing Phone Numbers:", df["Phone_Number"].isna().sum())
```

```python
print("Duplicate Rows Before Removal:", df.duplicated().sum())
```

```python
original_df = pd.read_csv(file_path)

duplicate_indices = original_df[original_df.duplicated()].index

print("Original Duplicate Rows:", len(duplicate_indices))

df = df.drop(index=duplicate_indices)

print("Rows After Removing Original Duplicates:", len(df))
print("Remaining Rows:", df.shape[0])
```

```python
print("Final Dataset Shape:", df.shape)

print("\nMissing Values in Each Column:")
display(df.isnull().sum())

print("\nGender Categories:")
print(df["Gender"].value_counts())

print("\nAge Range:")
print(df["Age"].min(), "to", df["Age"].max())

print("\nPurchase Amount Range:")
print(df["Purchase_Amount"].min(), "to", df["Purchase_Amount"].max())

print("\nDuplicate Rows Remaining:", df.duplicated().sum())
```

## Missing Values Handling – Strategy and Justification

| Column | Missing Value Strategy | Reason |
|---|---|---|
| Customer_ID | Kept as missing | Customer IDs were not available, so no artificial IDs were created. |
| Gender | Filled with "Unknown" | Missing gender information was not guessed. |
| Age | Median imputation | Median is less affected by extreme values than the mean. |
| City | Filled with "Unknown" | Missing city information was not assumed. |
| Last_Purchase_Date | Kept as missing | No reliable date was available to replace missing values. |
| Purchase_Amount | Invalid negative values handled; remaining missing values filled with median | Median provides a robust replacement for missing monetary values. |
| Feedback_Score | Median imputation | Median was used to fill missing feedback scores. |
| Phone_Number | Invalid phone numbers set to missing | Only valid 10-digit phone numbers were retained. |
| Country | Filled with "Unknown" | Missing country information was not assumed. |

**Note:** Columns with no missing values did not require imputation. All decisions were made to avoid inventing unreliable customer information.

## Outlier Detection and Treatment

The IQR method was applied to the Purchase_Amount column.

- Q1 = 12,295
- Q3 = 37,130
- IQR = 24,835
- Upper Bound = 74,382.5

Values above the upper bound were capped at 74,382.5. Negative purchase amounts were treated as invalid and set to missing before median imputation.

## Duplicate Removal

- Original dataset contained 15 exact duplicate rows.
- After data standardisation and missing-value handling, some additional rows became identical.
- A total of 171 duplicate rows were removed during the cleaning process.
- Final dataset contains 0 duplicate rows.

This step helped maintain data consistency and avoid repeated records.
